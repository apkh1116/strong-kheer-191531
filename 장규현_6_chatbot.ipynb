나만의 AI 챗봇 만들기

## 활동 1. 규칙 기반 챗봇
빈칸에 알맞은 핵심어를 넣으세요.


```python
print("학교생활 도우미 챗봇입니다.")
print("'종료'라고 입력하면 끝납니다.")

while True:
    question = input("\n나: ")

    if question == "종료":
        print("챗봇: 대화를 종료합니다.")
        break
    elif "급식" in question:
        print("챗봇: 급식에 관한 질문이군요.")
    elif "도서관" in question:
        print("챗봇: 도서관에 관한 질문이군요.")
    else:
        print("챗봇: 잘 모르겠습니다.")
```

```python
#[관찰]
#챗봇이 잘 대답한 질문 >> 급식과 도서관 키워드가 들어간 질문
#챗봇이 이해하지 못한 질문 >> 급식, 도서관 단어가 들어가지 않는 질문
#왜 이런 차이가 생겼을까? >> 급식과 도서관의 키워드 여부
```

## 활동 2. 학습 데이터 만들기
같은 의미이지만 표현이 다른 문장을 채우세요.


```python
questions = [
    "오늘 급식 뭐야", "이번주 수요일 점심 뭐 나와", "급식실이 어디야",
    "도서관 어디야", "도서관에 어떤 책 있어", "읽을 책 추천해 줄 수 있어",
    "진로 상담하고 싶어", "진로 쌤 어디 계시니", "어느 대학교를 가야해?"
]

labels = [
    "급식", "급식", "급식",
    "도서관", "도서관", "도서관",
    "진로", "진로", "진로"
]

print("학습 데이터 수:", len(questions))
```

## 활동 3. 문장을 숫자 특징으로 변환하기


#**입력** 문장: "오늘 급식" (공백 포함)

2글자 묶음 (2-gram):

"오늘", "늘 ", " 급", "급식"

3글자 묶음 (3-gram):

"오늘 ", "늘 급", " 급식"

**이렇게 조각난 ["오늘", "늘 ", " 급", "급식", "오늘 ", "늘 급", " 급식"]이라는 글자 패턴**

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(analyzer="char", ngram_range=(2, 3))
X = vectorizer.fit_transform(questions)

print(X.shape)
print(X.toarray())
```

```python
features = vectorizer.get_feature_names_out()
print("\n전체 특징 목록:\n", features)
```

## 활동 4. AI 학습시키기
`fit` 또는 `predict` 중 알맞은 것을 넣으세요.


```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X, labels)

print("학습 완료")
```

## 활동 5. 새로운 질문 분류하기


```python
question = input("질문: ")
X_new = vectorizer.transform([question])
result = model.predict(X_new)

print("AI 판단:", result[0])
```

## 활동 6. 나만의 챗봇 설계
범주는 최소 3개, 각 범주의 학습 문장은 최소 5개입니다.


```python
my_questions = [
    # 범주 1
    "이번 주 급식 어떤 거 나와?", "급식을 몇 번 다시받을 수 있어?", "선택 급식이 나오는 날을 알려줘", "이번주 급식은 학년 순서가 어떻게 돼?", "급식 시간이 언제부터 언제까지야?",

    # 범주 2
    "도서관 어디야", "어떤 종류의 도서가 있어?", "도서관 반경 1km 이내에 버스 정류장이 몇 개나 있어??", "도서가 총 몇 권이나 있어?", "책 반납 기간을 알려줘",

    # 범주 3
    "1등급대 성적으로 갈 수 있는 대학은 어디일까?", "생기부에 적히면 좋은 문장 알려줘", "주제 참구 예시 알려줘", "인서울 컷은 어디야", "내가 가고 싶은 학과는 어느 대학에 있는지 알고싶어",
]

my_labels = [
    "급식", "급식", "급식", "급식", "급식",
    "도서관", "도서관", "도서관", "도서관", "도서관",
    "진로", "진로", "진로", "진로", "진로"]
```

## 활동 7. 나만의 모델 학습


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
my_vectorizer = TfidfVectorizer(analyzer="char", ngram_range=(2, 3))
my_X = my_vectorizer.fit_transform(my_questions)

my_model = LogisticRegression(max_iter=1000)
my_model.fit(my_X, my_labels)

print("나만의 AI 모델 학습 완료!")
```

## 활동 8. 테스트하고 오류 찾기


### 최종 미션
친구가 학습 데이터에 없는 질문을 5개 입력합니다.  
정답 수를 기록하고, 틀린 질문을 분석하여 데이터를 추가한 뒤 다시 학습하세요.

```python
#질문1 >>어제 나왔던 급식은 뭐야?
#질문2 >>도서관에 책은 몇권 까지 빌릴 수 있어?
#질문3 >>급식실은 몇층에 있어?
#질문4 >>나의 적성에 맞는 대학을 추천해줘
#질문5 >>급식을 여러번 받을 수 있어?
#정답수 >> 4
#틀린 질문 분석 >> 새로운 질문: 급식을 여러번 받을 수 있어?/예측 범주: 도서관/
#모델 점수: 38.2 %에서, 모델 점수가 낮게 나온 것으로 보아, 평준화된 수준의 일부 질문에서도 챗봇이 잘 분류하지 못하고 분야를 잘못 내놓는 것으로 보인다.
```

```python
test_question = input("새로운 질문: ")
test_X = my_vectorizer.transform([test_question])

prediction = my_model.predict(test_X)[0]
probabilities = my_model.predict_proba(test_X)[0]

print("예측 범주:", prediction)
print("모델 점수:", round(max(probabilities) * 100, 1), "%")
```

## 활동9. 최종 챗봇

```python
print("=" * 40)
print("      학교생활 AI 챗봇")
print("=" * 40)
print("궁금한 것을 질문하세요.")
print("'종료'를 입력하면 프로그램이 종료됩니다.")

while True:
    question = input("\n나: ")
    if question == "종료":
        print("AI: 다음에 또 만나요!")
        break

    X_new = my_vectorizer.transform([question])
    intent = my_model.predict(X_new)[0]
    probabilities = my_model.predict_proba(X_new)[0]
    confidence = max(probabilities)

    print("분류 결과:", intent)
    print("모델 점수:", round(confidence * 100, 1), "%")
    if intent == "급식":
        print("AI: 급식에 관한 질문이군요!")
    elif intent == "도서관":
        print("AI: 도서관은 본관 2층에 있습니다.")
    elif intent == "진로":
        print("AI: 진로 고민이 있군요. 관심 있는 분야부터 생각해볼까요?")
```

**환각(Hallucination)이란?**

AI가 자신이 잘 모르는 내용임에도 불구하고, 마치 사실인 것처럼 당당하게 거짓 정보를 지어내어 답변하는 현상

```python
#1. 환각이 일어 나는 이유는? >> 대답하기 어려운 것에 대한 피드백, '대답하기 어렵습니다'를 출력하도록 하는 보완점이 부족하여, 어떻게든 답을 만들도록 하는 것이 이유라고 생각한다.
#2. 환각을 차단할 수 있는 방법은? >> 모르는 내용은 모른다고 응답하거나 인터넷을 검색하게 하는 경우를 학습시켜 환각을 막는다.
#3. 답변을 다양하게 만들 수 있는 방법은? >> 질문처럼 응답 문장을 학습시켜 질문에 맞는 단어를 분류하고 조합하며 다양한 문장을 형성시키게 한다.
#4. 최종챗봇이 질문을 얼마나 정확하게 분류하는지 평가 >> 오히려 분류하기 쉬운 질문도 잘 분류하지 못 하는 경향이 있다.
```
