## [문제] 당도와 아삭함을 값으로 과일, 단백질, 채소를 분류하는식품 분류 신경망 모델 설계하기(134p)

## [1단계] 데이터 탐색하기(food_nn.csv)

```python
#데이터 살펴보기
import pandas as pd
df=pd.read_csv('food_nn.csv')
df.head()
#class>>
```

## [2단계] 신경망 모델 설계하기

```python
#[2-1] X( 독립 변수)와 y(종속 변수  ) 지정
X = df[['sweet', 'crunchy']]
y=df['class']
```

```python
#[2-2] 신경망 모델 (keras) 라이브러리로 구현
from keras.models import Sequential
from keras.layers import  Dense
#층연결
model = Sequential()
#층구조>> 입력층(2)- 은닉층(32)-은닉층(16)-출력층(3)
model.add(Dense(32, input_dim=2, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(3, activation='softmax'))
#활성화함수>> 입력된 값의 가중합을 비선형으로 전달되도록 하는 함수이다.
```

```python
#[2-3] 신경망 모델 환경설정 >> cpmoile
#손실함수 >> 예측값과 실제 정답 간의 차이를 측정하는 함수
#최적화 함수 >> 손실함수의 값을 최소화하는 방향으로 신경망의 가중치 값을 조정하는 함수이다.
#평가지표 >> 정확도
#sparse_categorical_crossentropy
model.compile(loss= 'sparse_categorical_crossentropy',
              optimizer='adam', metrics=['accuracy'])
```

```python
#[2-4] 신경망 모델 학습 >>
model.fit(X, y, epochs=500)
```

## [3단계] 신경망 모델 평가 및 예측

```python
#모델 평가하기 >> evaluate
model.evaluate(X, y)
```

```python
#모델 예측하기 >> predict
model.predict(X)
```
